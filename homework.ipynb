```python
### WAP TO REVERSE GIVEN STRING
s = "instagram"
reverse = " "
for ch in s:
    reverse = ch + reverse
    print(reverse)
```

```python
### WAP TO COUNT  VOWELS IN GIVEN STRING
s = "instagram"
count = 0
for ch in s:
    if ch in "aeiou":
      count += 1
    print(count)
```

```python
### WAP TO COUNT CONSONANTS IN GIVEN STRING 
s = "instagram"
count = 0
for ch in s:
    if ch.isalpha( ) and ch not in "aeiou":
        count += 1
        print(count)
```

```python
### WAP TO CHECK GIVEN STRING IS PALINDROME OR NOT 
s = "madam"
reverse = ""

for ch in s:
    reverse = ch + reverse
if s == reverse:
    print("Palindrome")
else:
     print("Not Palindrome")   
```
